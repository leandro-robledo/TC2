### Bonus

Para la simulacion se tomaron los valores de Za=1, Zb=2, Zc=6, de forma de poder diferenciar bien los resultados obtenidos
Aqui se simulo los parametros Z de la red Pi, segun los calculos:

$Z11= \frac{ZaZc+ZaZb}{Za+Zb+Zc}=0.88=-1.023_{db}$

$Z12= \frac{ZaZb}{Za+Zb+Zc}=0.22=-13.064_{db}$

$Z21= \frac{ZaZb}{Za+Zb+Zc}=0.22=-13.064_{db}$

$Z22= \frac{ZbZc+ZaZb}{Za+Zb+Zc}=1.55=3.8377_{db}$
![valoresZ_Pi](Zparametros_Pi.PNG)

Simulacion los parametros Y de la red Pi, segun los calculos:

$Y11= Ya+Yb$

$Y12= -Yb$

$Y21= -Yb$

$Y22= Yb+Yc$


![valoresY_Pi](Yparametros_Pi.PNG)

Simulacion los parametros Z de la red Tee, segun los calculos:

$Z11= Za+Zb=3=9.54_{db}$

$Z12= Zb=2=6.02_{db}$

$Z21= Zb=2=6.02_{db}$

$Z22= Zc+Zb=8=18.06_{db}$

![valoresZ_Tee](Zparametros_Tee.PNG)

Simulacion los parametros Y de la red Tee, segun los calculos:

$Y11= \frac{YaYb+YaYc}{Ya+Yb+Yc}$

$Y12= \frac{-YaYc+Yc^2}{Ya+Yb+Yc}$

$Y21= \frac{- YaYc+Ya^2}{Ya+Yb+Yc} $

$Y22= \frac{YaYb+YbYc}{Ya+Yb+Yc}$


![valoresY_Pi](Yparametros_Pi.PNG)

Simulacion los parametros Z del laticce, segun los calculos:

$Z11= \frac{Zb+Za}{2}=1.5=3.52_{db}$

$Z12= \frac{Zb-Za}{2}=0.5=-6.02_{db}$

$Z21= \frac{Zb-Za}{2}=0.5=-6.02_{db}$

$Z22= \frac{Zb+Za}{2}=1.5=3.52_{db}$

![valoresZ_laticce](Zparametros_lattice.PNG)

Simulacion los parametros Y del laticce, segun los calculos:

$Y11= \frac{Za^2+2ZaZb+Zb^2}{2(ZaZb^2-Za^2Zb)}$

$Y12= \frac{Za+Zb}{2ZaZb} $

$Y21= \frac{Za+Zb}{2ZaZb} $

$Y22= \frac{Za^2+2ZaZb+Zb^2}{2(ZaZb^2-Za^2Zb)}$

![valoresY_laticce](Yparametros_lattice.PNG)
