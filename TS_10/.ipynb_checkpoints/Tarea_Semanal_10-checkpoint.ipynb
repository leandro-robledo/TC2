Para sacar la impedancia del circuito, primero se desarrolla el filtro deseado el cual comienza con la obtencion de los coeficientes del denominador del filtro chebyshev pedido. Una vez obtenido la transferencia pedida es igual a el modulo al cuadrado del coeficiente S21, el cual por definicion nos permite obtener el valor de S11, el cual finalmente nos permite llegar al valor de la impedancia de entrada Z1.

### Analisis de A <a id="S11_respuesta"></a>

Con la simulacion obtenida anteriormente se pueden obtener los siguientes valores:

S11 en centro de la banda de paso = -9.5db

S11 en frecuencia de corte = -2.57db

S11 en transición y centro de la banda de detenida = 0db 

Esto nos indica que el coeficiente de reflexion en la entrada en las frecuencias bajas, es decir por debajo de la frecuencia de corte, son muy pocas, menores a la mitad de la potencia entrgada. Por otro lado en la frecuencia de corte este valor enpieza a incrementar llegando a ser cercano a la los 3db de atenuacion de lo esperado en una frecuencia de corte. Finalmente en la banda de stop el coeficiente S11 tiende a 0db lo que nos dice que toda la onda es reflejada en su totalidad. Esto refiere correctamente al circuito, que al ser un pasabajo, en la banda de paso el coeficiente de reflexion es el menor a lo largo de todas las frecuencias. Por otro lado en la banda de stop del filtro la onda es completamente reflejada por lo que no hay potencia transmitida a la salida, lo que se espera de un filtro pasabajos en las frecuencias altas.

### NANO VNA

NanoVNA V2 (S-A-A-2) es un analizador de red vectorial T/R de 4 GHz de bajo costo desarrollado por HCXQS en colaboración con OwOComm, el cual no utiliza armonicos para las mediciones y logra una mayor repetibilidad de las mediciones.

Algunas de sus mejores caracteristicas:

Menor deriva de temperatura: las últimas versiones de NanoVNA V2 incluyen una calibración interna automática que corrige los errores de deriva
Alto rango dinámico de 90 dB: NanoVNA V2 Plus4 agrega blindaje mejorado, logra resultados similares a los VNA profesionales y permite medir duplexores.
Exportación de parámetros S; parámetros del kit de calibración personalizados & métodos de calibración avanzados (usando el software NanoVNA-QT)
TDR (reflectometría en el dominio del tiempo) permite medir la longitud y las discontinuidades del cable
Pantalla táctil de 4 pulgadas: se puede usar de forma independiente o con una PC (usando el software NanoVNA-QT o NanoVNA-Saver)

Es extremadamente util cuando se requiere medir los siguientes parametros:

El NanoVNA mide lo siguiente:

Coeficiente de reflexión: S11

Coeficiente de transmisión: S21

Se pueden mostrar los siguientes elementos que se pueden calcular a partir de estos:

Pérdida de devolución

Pérdida de inserción

Impedancia compleja

Resistencia

Reactancia

ROE

Si bien el DSO y el VNA no son idealmetne comparables al no cumplir exactametne la misma funcion a la hora de realizar mediciones, siendo el primero mas enfocado a las formas de ondas y transitorios de señales en el tiempo el VNA se emplea para la analisis de redes y elementos que la caracterizan. 

El DSO tiene un ancho de banda: 70, 100 o 150 MHz, según modelo, mientras que el del VNA es de 50 kHz–4,4 GHz.

El DSO tiene una Impedancia de entrada: $1M\Omega$ en paralelo con unos 15pF, mientras que el del VNA es de $50\Omega$.

- Rango dinámico calibrado del VNA va de entre 70 dB a 1 GHz y 60 dB a 3 GHz mientras que el del DSO tiene una resolución vertical de 8 bits.

  Por ultimo las funciones principales del DSO serian: visualizar formas de onda, mediciones automáticas, comparaciones y FFT, mienrtas que en el VNA seria calcular parametros $S_{11}, S_{21}$, impedancia compleja, ROE, pérdidas y TDR.
